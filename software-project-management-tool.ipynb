```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Importing The Software Project Management Tools Dataset**

```python
df = pd.read_csv('../input/software-project-management-tools/SPMQA Data Visualization - Sheet1.csv')
```

**Data Preprocessing/Data Cleaning**

```python
df1 = df.drop(columns=['Unnamed: 5'])
df1.columns = ['Roll_no', 'Tool1', 'Tool2', 'Tool3', 'Final_verdict']
```

```python
df1.sample(5)
```

Converting All Strings To Upper Case

```python
df1['Roll_no']  = df1['Roll_no'].str.upper()
df1['Tool1'] = df1['Tool1'].str.upper()
df1['Tool2'] = df1['Tool2'].str.upper()
df1['Tool3'] = df1['Tool3'].str.upper()
df1['Final_verdict'] =df1['Final_verdict'].str.upper()
```

**Stripping All the Spaces**

```python
df1['Final_verdict'] = df1['Final_verdict'].str.strip()
df1['Final_verdict'].unique().shape
```

```python
df1.sample(5)
```

```python
df1['Final_verdict']= df1['Final_verdict'].replace('TRELLO.COM', 'TRELLO')
df1['Final_verdict'].unique()
```

```python
df1['Final_verdict'] = df1['Final_verdict'].replace('MICROSOFT PROJECTS', 'MICROSOFT PROJECT')
df1['Final_verdict'].unique()
```

```python
df1['Final_verdict'] = df1['Final_verdict'].replace('ZOHO', 'ZOHO PROJECT')
df1['Final_verdict'] = df1['Final_verdict'].replace('ZOHO PROJECTS', 'ZOHO PROJECT')
df1['Final_verdict'] = df1['Final_verdict'].replace('CLICKUP', 'CLICK UP')
df1['Final_verdict'].unique()
```

**Data Visualization**

```python
sns.countplot(df1['Final_verdict'])
plt.xticks(rotation=90)
df1['Final_verdict'].value_counts()
```

**Tool1 Data Preprocessing**

```python
df1['Tool1'] = df1['Tool1'].str.strip()
df1['Tool1'].unique()
```

```python
df1['Tool1'] = df1['Tool1'].replace('ZOHO', 'ZOHO PROJECT')
df1['Tool1'] = df1['Tool1'].replace('TRELLO.COM', 'TRELLO')
df1['Tool1'] = df1['Tool1'].replace('MONDAY', 'MONDAY.COM')
df1['Tool1'] = df1['Tool1'].replace('MICROSOFT PROJECTS', 'MICROSOFT PROJECT')
df1['Tool1'] = df1['Tool1'].replace('CLICKUP', 'CLICK UP')
df1['Tool1'].unique()
```

```python
sns.countplot(df1['Tool1'])
plt.xticks(rotation=90)
df1['Tool1'].value_counts()
```

**Tool2 Data Preprocessing**

```python
df1['Tool2'] = df1['Tool2'].str.strip()
df1['Tool2'] = df1['Tool2'].replace('MONDAY', 'MONDAY.COM')
df1['Tool2'] = df1['Tool2'].replace('ZOHO PROJECTS', 'ZOHO PROJECT')
df1['Tool2'] = df1['Tool2'].replace('ZOHO', 'ZOHO PROJECT')
df1['Tool2'] = df1['Tool2'].replace('ZOHOCRM', 'ZOHO PROJECT')
df1['Tool2'] = df1['Tool2'].replace('CLICKUP', 'CLICK UP')
df1['Tool2'].unique()
```

**Tool3 Data Preprocessing**

```python
df1['Tool3'] = df1['Tool3'].str.strip()
df1['Tool3'].unique()
```

```python
df1['Tool3'] = df1['Tool3'].replace('MONDAY', 'MONDAY.COM')
df1['Tool3'] = df1['Tool3'].replace('MICROSOFT PROJECTS', 'MICROSOFT PROJECT')
df1['Tool3'] = df1['Tool3'].replace('MS PROJECTS', 'MICROSOFT PROJECT')
df1['Tool3'] = df1['Tool3'].replace('ZOHO PROJECTS', 'ZOHO PROJECT')
df1['Tool3'] = df1['Tool3'].replace('ZOHO', 'ZOHO PROJECT')
df1['Tool3'] = df1['Tool3'].replace('CLICKUP', 'CLICK UP')
df1['Tool3'] = df1['Tool3'].replace('LIQUIDPLANNER', 'LIQUID PLANNER')
df1['Tool3'].unique()
```
